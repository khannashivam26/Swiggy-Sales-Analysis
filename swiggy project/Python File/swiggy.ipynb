```python
df['Avg ratings']=df['Avg ratings'].round(1)
```

```python
df['Price']=df['Price'].round(1)
```

```python
df.head()
```

```python
df.columns
```

```python
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df.columns=df.columns.str.title()
```

```python
df.columns
```

```python
df.head()
```

```python
def price_map(x):
    if x<200: 
        return "Budget"
    elif x>=200 and x <400:
        return "Medium-Range"
    else :
        return "Costly"

df['Price_Category']=df['Price'].apply(price_map)
```

```python
df[['Price','Price_Category']].head(8)
```

```python
df['Price_Category']=df['Price_Category'].replace('Medium-Range','Mid Value')
```

```python
df.head()
```

```python
df['Delivery_Time'].value_counts()
```

```python
df['Delivery_Time'].describe()
```

```python
def del_stats(x):
    if x<45:
        return "Fast"
    elif x>=45 and x<80 :
        return "Moderate"
    else: 
        return "Slow"

df['Delivery_Status']=df['Delivery_Time'].apply(del_stats)
```

```python
df[['Delivery_Time','Delivery_Status']].head()
```

```python
df['Avg_Ratings'].unique()
```

```python
def avg_rate(x):
    if x<=3:
        return "Low Rated"
    elif x>3 and x<=4:
        return "Avg Rated"
    else:
        return "High Rated"

df['Rating_Status']=df['Avg_Ratings'].apply(avg_rate)
```

```python
df.head()
```

```python
df['Primary_Cuisine']=df['Food_Type'].str.split(',').str[0].str.strip()
```

```python
df.head()
```

```python
df['Number_Of_Cuisines']=df['Food_Type'].str.split(',').apply(len)
```

```python
df.head()
```

```python
df['City'].value_counts()
```

```python
df['Total_Ratings'].unique()
```

```python
import numpy as np

df['Popularity_Score']=(df['Avg_Ratings'] * np.log1p(df['Total_Ratings'])).round(3)
```

```python
df.head()
```

```python
df['Popularity_Score'].unique()
```

```python
df['Popularity_Score'].describe()
```

```python
def pop_stat(score):
    if score <10:
        return "Less Popular"
    elif score >=10 and score <25:
        return "Average Popular"
    else:
        return "More Popular"

df['Popularity_Status']=df['Popularity_Score'].apply(pop_stat)
```
